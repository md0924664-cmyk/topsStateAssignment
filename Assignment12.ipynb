task 1

State the null and alternative hypotheses for testing whether a new cashback offer on a food delivery app (like Zomato) has increased the average number of daily orders compared to before the offer.

```python
import numpy as np

np.random.seed(42)  # for reproducible results

# Simulate 30 days of daily order counts, before and after the cashback offer
orders_before = np.random.normal(loc=500, scale=40, size=30)
orders_after = np.random.normal(loc=530, scale=40, size=30)

mean_before = np.mean(orders_before)
mean_after = np.mean(orders_after)

print(f"Mean daily orders BEFORE the cashback offer: {round(mean_before, 2)}")
print(f"Mean daily orders AFTER the cashback offer: {round(mean_after, 2)}")

null_hypothesis = (
    "H0: mu_after <= mu_before -> The cashback offer has NOT increased average daily orders."
)
alternative_hypothesis = (
    "H1: mu_after > mu_before -> The cashback offer HAS increased average daily orders."
)

print(f"\n{null_hypothesis}")
print(alternative_hypothesis)
```

**Explanation:** We simulate 30 days of order data before and after the cashback offer to make the hypotheses concrete rather than purely abstract. The null hypothesis (H0) always represents "no improvement" — that the true mean daily orders after the offer is less than or equal to before. The alternative hypothesis (H1) is what we want evidence for — that the mean has genuinely increased. Since we only care about an *increase*, this is framed as a one-tailed test (using "&gt;" rather than "≠"). Just observing that the sample mean after (530-ish) looks higher than before (500-ish) is not enough on its own — that's exactly why we need a formal hypothesis test (done in Task 3) to check if the difference is statistically significant.

task 2

Given the following scenario: A music streaming app claims their new playlist recommendation algorithm increases average listening time from 40 to 45 minutes. If you set up a hypothesis test, identify which type of error (Type I or Type II) would occur if you wrongly conclude the algorithm works when it actually doesn't, and explain in one sentence why.

```python
import numpy as np
from scipy import stats

np.random.seed(1)

# Simulate a scenario where the algorithm actually has NO real effect (true means are equal)
# so that any "significant" result we find would be a false alarm
old_listening_time = np.random.normal(loc=40, scale=6, size=50)
new_listening_time = np.random.normal(loc=40, scale=6, size=50)  # same true mean as old

t_stat, p_value = stats.ttest_ind(new_listening_time, old_listening_time)
alpha = 0.05

print(f"Sample mean (old algorithm): {round(np.mean(old_listening_time), 2)} minutes")
print(f"Sample mean (new algorithm): {round(np.mean(new_listening_time), 2)} minutes")
print(f"t-statistic = {round(t_stat, 4)}, p-value = {round(p_value, 4)}")

if p_value < alpha:
    conclusion = "Reject H0 -> conclude the new algorithm increases listening time"
else:
    conclusion = "Fail to reject H0 -> not enough evidence the new algorithm helps"

print(f"Test conclusion: {conclusion}")

error_type = "Type I Error (False Positive)"
explanation = (
    "If the test above rejects H0 even though the true means were set equal (no real effect), "
    "that would be a Type I error, because we rejected a true null hypothesis - concluding the "
    "algorithm works purely due to random sampling noise, not a real improvement."
)

print(f"\nRelevant error type for this question: {error_type}")
print(explanation)
```

**Explanation:** To make this concrete, we simulate "new" and "old" listening times drawn from the *same* true mean (40 minutes) using `scipy.stats.ttest_ind`, so we know for certain the algorithm has no real effect in this simulation. If the resulting p-value still comes out below alpha=0.05 and we reject H0, that rejection would be a **Type I error** — wrongly concluding the algorithm works when it actually doesn't. (Run the cell again with a different seed and you'll occasionally see this happen purely by chance, which is exactly the ~5% false-positive rate that alpha=0.05 allows for.)

task 3

Suppose you run an A/B test on a shopping app (like Flipkart) and get a p-value of 0.03 when testing if a new discount banner increases purchases. At a 95% confidence level, state whether you reject or fail to reject the null hypothesis and explain what this means in the context of the app.

```python
import numpy as np
from scipy import stats

np.random.seed(7)

# Simulate an A/B test: control group (no banner) vs treatment group (with discount banner)
control_purchases = np.random.normal(loc=200, scale=30, size=100)     # no banner
treatment_purchases = np.random.normal(loc=212, scale=30, size=100)   # with banner

t_stat, p_value_simulated = stats.ttest_ind(treatment_purchases, control_purchases)

print(f"Control group mean purchases: {round(np.mean(control_purchases), 2)}")
print(f"Treatment group mean purchases: {round(np.mean(treatment_purchases), 2)}")
print(f"Simulated t-statistic = {round(t_stat, 4)}, simulated p-value = {round(p_value_simulated, 4)}")

# Now apply the decision rule using the p-value actually given in the question (0.03)
p_value = 0.03
confidence_level = 0.95
alpha = 1 - confidence_level

print(f"\nGiven p-value = {p_value}")
print(f"Confidence level = {confidence_level * 100}%, alpha = {round(alpha, 2)}")

if p_value < alpha:
    decision = "Reject the null hypothesis"
else:
    decision = "Fail to reject the null hypothesis"

print(f"Decision: {decision}")
```

**Explanation:** The simulated A/B test above shows how a p-value like this would actually arise from real purchase data using `scipy.stats.ttest_ind`. Applying the decision rule to the p-value given in the question (0.03): at a 95% confidence level, alpha = 1 - 0.95 = 0.05. Since 0.03 &lt; 0.05, we **reject the null hypothesis**. In the context of the app, this means there is statistically significant evidence that the new discount banner increases purchases — the observed difference is unlikely to be due to random chance alone, so Flipkart can reasonably roll out the banner.

task 4

For a cricket app's push notification feature, you want to be 99% confident that sending match alerts increases user engagement. Explain what the 99% confidence level means in this context, and how it affects your decision to accept or reject the feature.

```python
import numpy as np
from scipy import stats

np.random.seed(3)

# Simulate engagement scores (e.g., minutes active in app) for users who received match alerts
engagement_with_alerts = np.random.normal(loc=25, scale=6, size=200)

confidence_level = 0.99
alpha = 1 - confidence_level

# Build a 99% confidence interval for the true mean engagement using scipy
sample_mean = np.mean(engagement_with_alerts)
sample_sem = stats.sem(engagement_with_alerts)  # standard error of the mean
ci_low, ci_high = stats.t.interval(confidence_level, df=len(engagement_with_alerts) - 1,
                                    loc=sample_mean, scale=sample_sem)

print(f"Sample mean engagement (minutes) = {round(sample_mean, 2)}")
print(f"Confidence level = {confidence_level * 100}%, alpha = {round(alpha, 2)}")
print(f"99% Confidence Interval for true mean engagement = ({round(ci_low, 2)}, {round(ci_high, 2)})")

explanation = (
    "A 99% confidence level means that if this test were repeated many times, about 99% of the "
    "confidence intervals we build would contain the true average engagement level, leaving only "
    "a 1% chance of drawing the wrong conclusion (a Type I error) about whether match alerts "
    "increase engagement. Because alpha is so small (0.01), the app team needs a p-value below "
    "0.01 - a much stricter bar than the usual 0.05 - before they will accept that push "
    "notifications genuinely improve engagement and roll the feature out to everyone."
)

print(f"\n{explanation}")
```

**Explanation:** Using `scipy.stats.t.interval`, we build an actual 99% confidence interval around the simulated mean engagement score, which makes the abstract idea of "99% confidence" concrete: it's a range we're 99% sure contains the true average engagement of users receiving match alerts. Because this is stricter than the common 95% level, the significance level alpha drops to 0.01, meaning the observed improvement in engagement has to be very convincing (p-value below 0.01) before the cricket app team would accept and ship the push notification feature — reducing the risk of rolling out a feature that doesn't actually help, at the cost of needing stronger evidence.
