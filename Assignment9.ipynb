task 1

Write a Python script that generates 1000 random numbers following a normal distribution with mean 500 and standard deviation 100, then plot the resulting distribution using matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)  # for reproducible results

mean = 500
std_dev = 100
num_samples = 1000

sales_amounts = np.random.normal(loc=mean, scale=std_dev, size=num_samples)

print(f"Generated {num_samples} random numbers")
print(f"Sample mean: {round(np.mean(sales_amounts), 2)}")
print(f"Sample standard deviation: {round(np.std(sales_amounts), 2)}")

plt.figure(figsize=(8, 5))
plt.hist(sales_amounts, bins=30, color="skyblue", edgecolor="black")
plt.axvline(mean, color="red", linestyle="--", label=f"Mean = {mean}")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.title("Simulated Normal Distribution of Sales Amounts")
plt.legend()
plt.show()
```

**Explanation:** `np.random.normal(loc=500, scale=100, size=1000)` generates 1000 random values drawn from a normal distribution centered at 500 with a standard deviation of 100. The histogram shows the classic bell-shaped curve, with most values clustering close to 500 and fewer values as we move further away in either direction. The sample mean and standard deviation we calculate should be close to 500 and 100 respectively, confirming the simulation matches the specified distribution.

task 2

Given an array of daily step counts from a fitness app, calculate the Z-score for a specific day where the user walked 9500 steps. Assume the mean is 8000 and standard deviation is 1200.

**Hint:** Use the formula Z = (X - μ) / σ.

```python
import numpy as np

x = 9500          # the specific day's step count
mean = 8000       # mean (mu)
std_dev = 1200    # standard deviation (sigma)

z_score = np.divide(x - mean, std_dev)

print(f"Step count (X) = {x}")
print(f"Mean (mu) = {mean}")
print(f"Standard deviation (sigma) = {std_dev}")
print(f"Z-score = ({x} - {mean}) / {std_dev} = {round(z_score, 4)}")
```

**Explanation:** The Z-score tells us how many standard deviations a value is above or below the mean, using the formula Z = (X - μ) / σ. Here, Z = (9500 - 8000) / 1200 = 1.25. This means walking 9500 steps is 1.25 standard deviations above the average daily step count of 8000 — a noticeably active day compared to the user's usual pattern, but not extremely unusual (within about 2 standard deviations of the mean).

task 3

Plot a standard normal curve (mean 0, standard deviation 1) using matplotlib, and shade the area between Z = -1 and Z = 1 to visualize the probability of values within one standard deviation.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generate x values across the standard normal curve
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x, loc=0, scale=1)

# Values between Z = -1 and Z = 1, for shading
x_shade = np.linspace(-1, 1, 500)
y_shade = norm.pdf(x_shade, loc=0, scale=1)

# Probability of being within one standard deviation of the mean
prob_within_1sd = norm.cdf(1) - norm.cdf(-1)

plt.figure(figsize=(8, 5))
plt.plot(x, y, color="navy", label="Standard Normal Curve")
plt.fill_between(x_shade, y_shade, color="skyblue", alpha=0.6, label="Z = -1 to Z = 1")
plt.xlabel("Z-score")
plt.ylabel("Probability Density")
plt.title("Standard Normal Distribution (Mean = 0, Std Dev = 1)")
plt.legend()
plt.show()

print(f"Probability that Z is between -1 and 1 = {round(prob_within_1sd, 4)}")
```

**Explanation:** The standard normal curve has mean 0 and standard deviation 1, and `scipy.stats.norm.pdf` gives us the height of the curve at each point for plotting. The shaded region between Z = -1 and Z = 1 represents the values within one standard deviation of the mean. Using `norm.cdf(1) - norm.cdf(-1)`, this comes out to approximately 0.6827, confirming the well-known "68-95-99.7 rule": about 68% of values in a normal distribution fall within one standard deviation of the mean.

task 4

You are analyzing the delivery times (in minutes) for a food delivery app like Zomato. Given a normal distribution with mean 30 and standard deviation 5, write code to find the probability that a randomly selected delivery takes less than 25 minutes.

**Hint:** Use scipy.stats.norm.cdf().

```python
from scipy.stats import norm

mean_delivery_time = 30   # minutes
std_dev_delivery_time = 5 # minutes
x = 25                    # the delivery time we're interested in

# P(X < 25) using the cumulative distribution function
prob_less_than_25 = norm.cdf(x, loc=mean_delivery_time, scale=std_dev_delivery_time)

print(f"Mean delivery time = {mean_delivery_time} minutes")
print(f"Standard deviation = {std_dev_delivery_time} minutes")
print(f"P(Delivery time < {x} minutes) = {round(prob_less_than_25, 4)}")
```

**Explanation:** `norm.cdf(x, loc=mean, scale=std_dev)` gives the cumulative probability P(X < x) for a normal distribution. Here, with mean 30 minutes and standard deviation 5 minutes, P(Delivery time < 25 minutes) ≈ 0.1587. This makes sense because 25 minutes is exactly one standard deviation below the mean (Z = -1), and from the standard normal curve we know only about 15.87% of the distribution lies below that point — so roughly 1 in 6 deliveries would be expected to arrive in under 25 minutes.
