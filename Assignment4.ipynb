task 1

Given the daily step counts of a user for the last 7 days as [6500, 7200, 8000, 6000, 9000, 7500, 8200], calculate the range of steps and explain what this tells you about the user's activity pattern.

```python
import numpy as np
import pandas as pd

steps = np.array([6500, 7200, 8000, 6000, 9000, 7500, 8200])

maximum = np.max(steps)
minimum = np.min(steps)
step_range = np.ptp(steps)

print("Maximum steps:", maximum)
print("Minimum steps:", minimum)
print("Range:", step_range)
```

task 2

For a list of 5 food delivery order amounts on Zomato: [250, 400, 350, 300, 500], write code to compute the variance and standard deviation of the order amounts.

```python
import numpy as np

orders = np.array([250, 400, 350, 300, 500])

variance = np.var(orders)
standard_deviation = np.std(orders)

print("Variance:", variance)
print("Standard Deviation:", standard_deviation)
```

```python
import pandas as pd
orders = pd.Series([250, 400, 350, 300, 500])

print("Variance:", orders.var(ddof=0))
print("Standard Deviation:", orders.std(ddof=0))
```

task 3

You have the monthly mobile data usage (in GB) for a user over 6 months: [3.2, 2.8, 3.5, 3.0, 2.9, 3.1]. Calculate the coefficient of variation (CV) and interpret whether the user's data usage is consistent or highly variable.<br><br><em><strong>Hint:</strong> CV = (Standard Deviation / Mean) × 100%</em>

```python
import numpy as np

data_usage = np.array([3.2, 2.8, 3.5, 3.0, 2.9, 3.1])

mean = np.mean(data_usage)
std = np.std(data_usage)

cv = (std / mean) * 100

print("Mean:", mean)
print("Standard Deviation:", std)
print("Coefficient of Variation:", cv, "%")
```

Suppose you are analyzing the number of likes on 10 recent Instagram posts: [120, 150, 130, 110, 170, 140, 135, 160, 125, 155]. Compute the standard deviation and compare it with the mean to comment on the consistency of engagement
