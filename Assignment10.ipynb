task 1

Create a Python script that generates a random sample of 20 users from a list of 200 user IDs, simulating how a music app like Spotify might select users for a feedback survey.

```python
import numpy as np

np.random.seed(42)  # for reproducible results

# Simulate 200 Spotify user IDs
user_ids = [f"user_{i}" for i in range(1, 201)]

# Randomly select 20 users without replacement (simple random sampling)
sample_size = 20
random_sample = np.random.choice(user_ids, size=sample_size, replace=False)

print(f"Total users: {len(user_ids)}")
print(f"Sample size: {sample_size}")
print("\nRandomly selected users for the feedback survey:")
print(random_sample)
```

**Explanation:** Simple random sampling means every user has an equal chance of being picked, with no bias toward any particular group. Here, `np.random.choice(user_ids, size=20, replace=False)` selects 20 unique users out of the 200 total, `replace=False` ensuring no user is picked twice. This is a quick, unbiased way for Spotify to choose a representative group of users to send a feedback survey to.

task 2

Given an array of 1000 Flipkart orders labeled by city (e.g., 'Ahmedabad', 'Surat', 'Vadodara'), write code to generate a stratified sample of 60 orders, ensuring each city is proportionally represented in the sample.

**Hint:** Calculate the proportion of orders per city and sample accordingly from each group.

```python
import numpy as np
import pandas as pd

np.random.seed(42)  # for reproducible results

# Simulate 1000 Flipkart orders labeled by city (unequal distribution across cities)
cities = ["Ahmedabad", "Surat", "Vadodara"]
city_weights = [0.5, 0.3, 0.2]  # Ahmedabad gets more orders than Surat and Vadodara

order_cities = np.random.choice(cities, size=1000, p=city_weights)
orders_df = pd.DataFrame({"order_id": range(1, 1001), "city": order_cities})

print("Total orders per city:")
print(orders_df["city"].value_counts())

# Stratified sampling: sample from each city in proportion to its share of the total
total_sample_size = 60
stratified_sample_parts = []

for city in cities:
    city_orders = orders_df[orders_df["city"] == city]
    city_proportion = len(city_orders) / len(orders_df)
    city_sample_size = round(city_proportion * total_sample_size)
    city_sample = city_orders.sample(n=city_sample_size, random_state=42)
    stratified_sample_parts.append(city_sample)

stratified_sample = pd.concat(stratified_sample_parts).reset_index(drop=True)

print(f"\nStratified sample size: {len(stratified_sample)}")
print("\nSample count per city:")
print(stratified_sample["city"].value_counts())
```

**Explanation:** Stratified sampling divides the population into groups (strata) — here, cities — and samples from each group in proportion to its share of the total population. We calculate each city's proportion of the 1000 orders, then multiply by the target sample size (60) to get how many orders to draw from that city. This ensures Ahmedabad, Surat, and Vadodara are all represented in the sample at roughly the same ratio as they appear in the full data set, unlike simple random sampling which could accidentally over- or under-represent a smaller city.

task 3

Write a function to perform systematic sampling on a list of 500 Zomato restaurant reviews, selecting every 10th review to create a sample set.

```python
def systematic_sample(data, interval):
    """
    Selects every 'interval'-th item from the data list, starting at index 0.
    """
    return data[::interval]

# Simulate 500 Zomato restaurant reviews
reviews = [f"review_{i}" for i in range(1, 501)]

sampling_interval = 10
sample = systematic_sample(reviews, sampling_interval)

print(f"Total reviews: {len(reviews)}")
print(f"Sampling interval: every {sampling_interval}th review")
print(f"Sample size: {len(sample)}")
print("\nFirst 10 items in the sample:")
print(sample[:10])
```

**Explanation:** Systematic sampling picks items at a fixed interval (here, every 10th review) after choosing a starting point, rather than picking randomly each time. With 500 reviews and an interval of 10, this gives a sample of exactly 50 reviews spread evenly across the whole list. It's simple to implement and works well as long as there's no hidden repeating pattern in the data that lines up with the interval and introduces bias.

task 4

Suppose you want to estimate the average delivery time for Swiggy orders in your city. Explain with calculations how the sample size affects the accuracy of your estimate if you have access to 10,000 delivery records and can only sample 50, 200, or 500 records.

```python
import numpy as np

np.random.seed(42)  # for reproducible results

# Simulate the full population of 10,000 Swiggy delivery times (in minutes)
population_size = 10000
true_mean = 32
true_std = 8
population = np.random.normal(loc=true_mean, scale=true_std, size=population_size)

print(f"True population mean delivery time: {round(np.mean(population), 2)} minutes")
print(f"True population standard deviation: {round(np.std(population), 2)} minutes\n")

sample_sizes = [50, 200, 500]

for n in sample_sizes:
    sample = np.random.choice(population, size=n, replace=False)
    sample_mean = np.mean(sample)
    # Standard error tells us how much the sample mean is expected to vary from the true mean
    standard_error = np.divide(np.std(sample), np.sqrt(n))

    print(f"Sample size = {n}")
    print(f"  Sample mean = {round(sample_mean, 2)} minutes")
    print(f"  Standard error = {round(standard_error, 4)}")
    print(f"  Difference from true mean = {round(abs(sample_mean - true_mean), 2)} minutes\n")
```

**Explanation:** The standard error, calculated as (sample standard deviation) / sqrt(sample size), tells us how much a sample mean is likely to vary from the true population mean. As the sample size increases from 50 to 200 to 500, the standard error shrinks (since we're dividing by a larger sqrt(n)), meaning the sample mean becomes a more reliable estimate of the true delivery time. In the output above, the sample of 500 typically ends up closer to the true mean of 32 minutes than the sample of 50, illustrating the general rule: larger samples give more accurate and stable estimates, but with diminishing returns — going from 50 to 200 improves accuracy more noticeably than going from 200 to 500.

Modify the stratified sampling demo shown in class: instead of customer segments, use a dataset of IPL cricket fans grouped by favorite team, and generate a stratified sample of 40 fans for a survey, ensuring each team is represented proportionally.

**Constraint:** Do not copy the code from the demo—write your own logic for grouping and sampling.

```python
import random

random.seed(42)  # for reproducible results

# Simulate 600 IPL fans, each supporting one team (unequal fan base sizes, no pandas this time)
teams = ["CSK", "MI", "RCB", "KKR", "GT"]
team_fan_counts = {"CSK": 180, "MI": 150, "RCB": 140, "KKR": 80, "GT": 50}  # totals sum to 600

# Build a plain list of (fan_id, team) tuples instead of using groupby
fans = []
fan_id = 1
for team, count in team_fan_counts.items():
    for _ in range(count):
        fans.append((fan_id, team))
        fan_id += 1

total_fans = len(fans)
print(f"Total fans: {total_fans}")
print(f"Fans per team: {team_fan_counts}\n")

# Group fan IDs manually into a dictionary, without using pandas groupby
fans_by_team = {}
for fid, team in fans:
    fans_by_team.setdefault(team, []).append(fid)

# Perform proportional stratified sampling using random.sample on each team's fan list
target_sample_size = 40
stratified_sample = {}

for team, fan_list in fans_by_team.items():
    proportion = len(fan_list) / total_fans
    team_sample_size = round(proportion * target_sample_size)
    stratified_sample[team] = random.sample(fan_list, team_sample_size)

print("Stratified sample sizes per team:")
for team, sampled_ids in stratified_sample.items():
    print(f"  {team}: {len(sampled_ids)} fans sampled -> {sampled_ids}")

total_sampled = sum(len(v) for v in stratified_sample.values())
print(f"\nTotal fans sampled: {total_sampled}")
```

**Explanation:** This uses plain Python dictionaries and lists instead of pandas, so the grouping and sampling logic is written from scratch. Fans are first generated with a team label, then manually grouped into a `fans_by_team` dictionary using a loop (instead of `groupby`). For each team, the proportion of total fans is calculated and multiplied by the target sample size (40) to decide how many fans to draw from that team using `random.sample`. This keeps the sample proportional to each team's real fan base — CSK, with the largest fan base, contributes the most fans to the survey sample, while GT, with the smallest, contributes the fewest.
