task 1

Simulate the Central Limit Theorem in Python by taking 1000 random samples (size 30) from a skewed dataset (e.g., number of likes on Instagram posts), calculating the mean of each sample, and plotting the histogram of these sample means using matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)  # for reproducible results

# Simulate a skewed population: number of likes on Instagram posts
# Most posts get relatively few likes, a small number go viral -> right-skewed data
population_size = 100000
instagram_likes = np.random.exponential(scale=200, size=population_size)

# Plot the original skewed population
plt.figure(figsize=(8, 5))
plt.hist(instagram_likes, bins=50, color="salmon", edgecolor="black")
plt.xlabel("Number of Likes")
plt.ylabel("Frequency")
plt.title("Original Skewed Population: Instagram Post Likes")
plt.show()

# Take 1000 random samples of size 30 and calculate the mean of each sample
num_samples = 1000
sample_size = 30
sample_means = [np.mean(np.random.choice(instagram_likes, size=sample_size, replace=False))
                 for _ in range(num_samples)]

# Plot the histogram of sample means
plt.figure(figsize=(8, 5))
plt.hist(sample_means, bins=30, color="skyblue", edgecolor="black")
plt.axvline(np.mean(instagram_likes), color="red", linestyle="--", label="Population Mean")
plt.xlabel("Sample Mean (Likes)")
plt.ylabel("Frequency")
plt.title(f"Sampling Distribution of the Mean (n = {sample_size}, 1000 samples)")
plt.legend()
plt.show()

print(f"Population mean: {round(np.mean(instagram_likes), 2)}")
print(f"Mean of sample means: {round(np.mean(sample_means), 2)}")
```

**Explanation:** The population of Instagram likes is simulated with an exponential distribution, which is heavily right-skewed — most posts get relatively few likes, while a small number go viral with very high like counts. Despite this skew, when we repeatedly draw samples of size 30 and calculate their means 1000 times, the histogram of those sample means comes out roughly bell-shaped and centered near the true population mean. This is the Central Limit Theorem in action: the sampling distribution of the mean approaches normality even when the underlying data is not normal, as long as the sample size is reasonably large.

task 4

Modify the CLT simulation code to use a much smaller sample size (n=5) and compare the histogram of sample means to the one with n=30. Briefly describe what changes you observe in the shape of the distribution.

**Hint:** Run both simulations and compare the spread and normality of the histograms.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)  # for reproducible results

# Reuse the same skewed population of Instagram likes
population_size = 100000
instagram_likes = np.random.exponential(scale=200, size=population_size)

num_samples = 1000

# Sampling distribution with small sample size (n = 5)
small_sample_size = 5
sample_means_n5 = [np.mean(np.random.choice(instagram_likes, size=small_sample_size, replace=False))
                    for _ in range(num_samples)]

# Sampling distribution with larger sample size (n = 30), same as Task 1
large_sample_size = 30
sample_means_n30 = [np.mean(np.random.choice(instagram_likes, size=large_sample_size, replace=False))
                     for _ in range(num_samples)]

# Plot both histograms side by side for comparison
fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)

axes[0].hist(sample_means_n5, bins=30, color="orange", edgecolor="black")
axes[0].set_title(f"Sample Means (n = {small_sample_size})")
axes[0].set_xlabel("Sample Mean (Likes)")
axes[0].set_ylabel("Frequency")

axes[1].hist(sample_means_n30, bins=30, color="skyblue", edgecolor="black")
axes[1].set_title(f"Sample Means (n = {large_sample_size})")
axes[1].set_xlabel("Sample Mean (Likes)")

plt.tight_layout()
plt.show()

print(f"Std dev of sample means (n=5):  {round(np.std(sample_means_n5), 2)}")
print(f"Std dev of sample means (n=30): {round(np.std(sample_means_n30), 2)}")
```

**Explanation:** With a small sample size (n=5), the histogram of sample means is noticeably wider and still shows some right-skew, since averaging only 5 values isn't enough to fully cancel out the effect of a single unusually large "viral" value in the sample. With the larger sample size (n=30), the histogram becomes narrower (a smaller standard deviation, as shown in the printed output) and looks much closer to a smooth, symmetric bell curve. This matches CLT theory directly: the sampling distribution of the mean gets both more normal-shaped *and* less spread out (standard error shrinks as sample size grows) as the sample size increases.

The left plot confirms the population itself is strongly right-skewed with most values bunched near zero and a long tail. The right plot — the sampling distribution of the mean for samples of size 40 — comes out clearly bell-shaped and centered close to the population mean of 1, even though the original exponential distribution looked nothing like a normal curve. This reinforces the same conclusion as the earlier tasks: regardless of the shape of the original data, averaging over a reasonably large sample produces a sampling distribution that is approximately normal.
